### **Elaboración de listado de 10 áreas de estudio aleatorio**

Este código define una función llamada `get_random`, que selecciona aleatoriamente un elemento (en este caso, un área de investigación) de una lista en formato de texto. La función espera que cada área esté numerada y formateada de una manera específica (número, punto y espacio) al inicio de cada línea. Vamos a desglosar cada parte del código:

Código Python:

  	import random


1. `import random`: Importa el módulo `random`, que permite generar valores aleatorios. Aquí se usa para seleccionar un elemento aleatorio de una lista.

Código Python:


    def get_random(topics):
        """
        Obtiene un área de investigación aleatoria de la lista proporcionada.

        Args:
            topics: Una cadena de texto que contiene la información a buscar, donde se supone que cada área comienza con un número seguido de un punto y un espacio.

        Returns:
            Una cadena de texto que representa el dato buscado seleccionada aleatoriamente, o None si no se encuentra ninguna área válida.
        """

    
2. Definición de la función `get_random`: Esta función toma un argumento llamado `topics`, que es una cadena de texto con múltiples líneas, donde se espera que cada área de investigación esté enumerada siguiendo un formato específico.
       
3. `Docstring`: La cadena de documentación (`docstring`) explica el propósito de la función, los argumentos y lo que devuelve:
  * `topics`: Debe ser una cadena en la que cada línea potencialmente contiene un área de investigación. Las líneas válidas empiezan con un número, seguido de un punto (`.`) y un espacio.
  * `Return`: La función devuelve una cadena con un área de investigación seleccionada aleatoriamente, o `None` si no se encuentra ninguna área válida.

Código Python:

    potential_areas = []

4. `potential_areas`: Se define una lista vacía que almacenará las áreas de investigación válidas.

Código Python:

    for line in topics.split('\n'):
        if line.strip() and line.strip()[0].isdigit() and line.strip()[1] == '.':
            potential_areas.append(line.strip())

5. `Iteración y Filtrado`:
  * `for line in topics.split('\n')`: Divide topics en líneas separadas usando `split('\n')` y recorre cada línea.
    * Condición `if`:
      * `line.strip()`: Asegura que la línea no esté vacía.
      * `line.strip()[0].isdigit()`: Verifica si el primer carácter de la línea es un dígito (lo que significa que está numerada).
      * `line.strip()[1] == '.'`: Verifica si el segundo carácter es un punto (.), siguiendo el formato esperado.
      * Si la línea cumple todas estas condiciones, se agrega (sin espacios adicionales) a `potential_areas`.


Código Python:
    if potential_areas:
        return random.choice(potential_areas)
    else:
        return None

6. Devolución del Resultado:
  * `if potential_areas`: Verifica si la lista potential_areas contiene elementos.
    * `random.choice(potential_areas)`: Si `potential_areas` no está vacía, selecciona y devuelve un elemento aleatorio de la lista.
    * `else: return None`: Si no hay elementos en `potential_areas`, devuelve `None`, indicando que no se encontró ningún área válida.


**Ejemplo de uso**
Si `topics` contiene el siguiente texto:
markdown
1. Inteligencia Artificial
2. Ciencia de Datos
3. Seguridad Informática

Llamar a `get_random(topics)` podría devolver cualquier línea (como `"2. Ciencia de Datos"`) o `None` si no se encuentran líneas válidas.

**Resumen**
La función `get_random`:
* Filtra y selecciona líneas que cumplen un formato específico (numeración seguida de un punto y un espacio).
* Devuelve aleatoriamente una línea válida de la lista o `None` si no encuentra ninguna.

```python
import random

def get_random(topics):
    """
    Obtiene un área de investigación aleatoria de la lista proporcionada.

    Args:
        topics: Una cadena de texto que contiene la información a buscar, donde se supone que cada área comienza con un número seguido de un punto y un espacio.

    Returns:
        Una cadena de texto que representa el dato buscado seleccionada aleatoriamente, o None si no se encuentra ninguna área válida.
    """
    potential_areas = []
    for line in topics.split('\n'):
        if line.strip() and line.strip()[0].isdigit() and line.strip()[1] == '.':
            potential_areas.append(line.strip())

    if potential_areas:
        return random.choice(potential_areas)
    else:
        return None
```
