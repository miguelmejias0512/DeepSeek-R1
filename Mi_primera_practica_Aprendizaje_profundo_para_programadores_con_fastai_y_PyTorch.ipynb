````python
from duckduckgo_search import DDGS # Use DDGS class instead of ddg_images
from fastcore.all import *

def search_images(term, max_images=30):
    print(f"Searching for '{term}'")
    with DDGS() as ddgs:
        results = L(ddgs.images(keywords=term, max_results=max_images)).itemgot('image')
    return results
````

Resumen:
* El código busca imágenes relacionadas con `'forest'` y `'bird'`, descargándolas en directorios correspondientes.

* Pausa entre cada descarga para evitar sobrecargar los servidores de las imágenes.

````python
searches = 'forest','bird'
path = Path('bird_or_not')
from time import sleep

for o in searches:
    dest = (path/o)
    dest.mkdir(exist_ok=True, parents=True)
    download_images(dest, urls=search_images(f'{o} photo'))
    sleep(10)  # Pause between searches to avoid over-loading server
    download_images(dest, urls=search_images(f'{o} sun photo'))
    sleep(10)
    download_images(dest, urls=search_images(f'{o} shade photo'))
    sleep(10)
    resize_images(path/o, max_size=400, dest=path/o)
````

Este código es parte de un flujo de trabajo de aprendizaje automático utilizando la biblioteca `Fastai`. Se está creando un conjunto de datos (`dls`) de imágenes, aplicando transformaciones y dividiendo las imágenes en subconjuntos de entrenamiento y validación.

1. **Creación del DataBlock**

Código Python:
```
dls = DataBlock(
    blocks=(ImageBlock, CategoryBlock),
    get_items=get_image_files,
    splitter=RandomSplitter(valid_pct=0.2, seed=42),
    get_y=parent_label,
    item_tfms=[Resize(192, method='squish')]
)
```

Vamos a descomponer lo que ocurre en cada parte:

* **blocks=(ImageBlock, CategoryBlock)** `ImageBlock` Especifica que el conjunto de datos contendrá imágenes como entrada. `CategoryBlock` Especifica que las etiquetas o categorías de salida serán clases categóricas (por ejemplo, diferentes tipos de animales, plantas, etc.).
* **get_items=get_image_files** `get_image_files` Es una función que obtiene todos los archivos de imagen en el directorio dado. Este es el método para recolectar los ítems (imágenes) que serán parte del conjunto de datos.
* **splitter=RandomSplitter(valid_pct=0.2, seed=42)**
`RandomSplitter` Divide las imágenes en dos conjuntos, uno de entrenamiento y otro de validación, de manera aleatoria. `valid_pct=0.2` El 20% de las imágenes se utilizarán para validación, y el 80% para entrenamiento. `seed=42` Establece una semilla para asegurar que la división aleatoria sea reproducible (para que siempre se dividan de la misma manera en futuras ejecuciones del código).
* **get_y=parent_label** `parent_label` Función que obtiene la etiqueta de clase de cada imagen, basada en el nombre del directorio en el que se encuentra. Es útil cuando las imágenes están organizadas en carpetas que corresponden a sus categorías. Por ejemplo, si las imágenes de perros están en una carpeta llamada "dog", parent_label asigna "dog" como la etiqueta de esas imágenes.
* **item_tfms=[Resize(192, method='squish')]** `item_tfms` Especifica las transformaciones que se aplican a cada imagen antes de ingresarlas al modelo.
* **Resize(192, method='squish'):** Redimensiona las imágenes para que tengan un tamaño de 192 píxeles (en ambos ejes) usando el método `'squish'`, que ajusta la imagen exactamente a ese tamaño, deformando la imagen si es necesario.

2. **Generación de los dataloaders**

Código Python:
```
.dataloaders(path, bs=32)
```
* `.dataloaders(path, bs=32)`: Crea los dataloaders, que son los objetos que manejan la carga de datos en lotes durante el entrenamiento del modelo.
* `path`: Es el directorio que contiene las imágenes. Se pasa como argumento para que el dataloader sepa de dónde cargar los datos.
* `bs=32`:  Establece el tamaño del lote (`batch size`) en 32. Esto significa que las imágenes se cargan y procesan en lotes de 32.

3. **Visualización de un lote de imágenes**

Código Python:
```
dls.show_batch(max_n=6)
```

* `dls.show_batch(max_n=6)`: Muestra un lote de hasta 6 imágenes con sus correspondientes etiquetas. Es útil para inspeccionar visualmente si las imágenes y las etiquetas están correctamente alineadas.

El siguiente código es parte de un flujo de trabajo de `Fastai` que se utiliza para entrenar un modelo de visión por computadora basado en una red neuronal convolucional preentrenada, en este caso, `ResNet-18`. Vamos a desglosarlo paso a paso:

1. **Creación del modelo de aprendizaje (learner)**

Código Python:
```
learn = vision_learner(dls, resnet18, metrics=error_rate)
```

* `vision_learner`: Es una función de Fastai que crea un "aprendiz" o learner para tareas de visión por computadora, como la clasificación de imágenes. El learner es el objeto que encapsula el modelo, los datos, y la lógica de entrenamiento. Los argumentos de esta función son:
  * `dls`: El conjunto de datos, representado por los dataloaders que contienen los datos de entrenamiento y validación. En este caso, `dls` contiene imágenes y sus respectivas etiquetas, posiblemente creadas previamente usando algo como `DataBlock`.
  * `resnet18`: Es el modelo base de la familia ResNet, una red neuronal convolucional muy popular, preentrenada en el conjunto de datos ImageNet. Al usar `resnet18`, el código reutiliza este modelo preentrenado y ajusta sus pesos a la tarea específica (por ejemplo, identificar pájaros).
  * `metrics=error_rate`: Especifica que la métrica a monitorear durante el entrenamiento es la tasa de error (error_rate), que se calcula como 1 - precisión. Otras métricas como la precisión (`accuracy`) también podrían haberse usado.

2. **Ajuste fino del modelo (fine-tuning)**

Código Python:
```
learn.fine_tune(3)
```
* `fine_tune`: Esta función realiza el ajuste fino o fine-tuning del modelo preentrenado. El ajuste fino ajusta el modelo a la nueva tarea, en este caso, entrenando en el conjunto de datos `dls`, pero con el conocimiento previo del modelo preentrenado en otro conjunto de datos grande como ImageNet.

**¿Cómo funciona?**
* Al usar un modelo preentrenado como `resnet18`, el modelo ya tiene pesos optimizados para una tarea general de clasificación de imágenes. En lugar de entrenar el modelo desde cero (lo que sería costoso y menos eficiente), el ajuste fino reutiliza esos pesos y los adapta a la tarea específica que tiene el nuevo conjunto de datos (por ejemplo, detectar diferentes tipos de animales o flores).

**Argumento 3:**

* El número `3` indica que el modelo se entrenará durante 3 épocas (es decir, el conjunto de datos será procesado 3 veces durante el entrenamiento).
* Durante la primera parte del entrenamiento, generalmente sólo se entrenan las capas más altas (las más especializadas) del modelo, manteniendo las capas más básicas congeladas con los pesos preentrenados.
* Luego, se descongela el modelo completo y se entrena todas las capas para adaptar mejor los pesos a la tarea específica.

````python
learn = vision_learner(dls, resnet18, metrics=error_rate)
learn.fine_tune(3)
````

### **IV. Paso 03: Utilizando el Modelo.** ###

1. **Predicción del modelo**

Código Python:
```

```
* `learn.predict()`: Utiliza el modelo `learn` (previamente entrenado) para hacer una predicción sobre una imagen.
* `PILImage.create('/content/bird.jpg')`: Carga la imagen desde el archivo `'/content/bird.jpg'` usando la clase `PILImage` para convertirla en un formato que el modelo pueda interpretar.
* `learn.predict()`: Retorna tres valores
`is_bird` La etiqueta predicha por el modelo, que indica si la imagen es o no un pájaro (según las categorías en las que el modelo fue entrenado).
* `_`: Ignorado en este caso (es la representación en números de la clase predicha).
* `probs`: Probabilidades asociadas a cada clase predicha. Es una lista o tensor que contiene la probabilidad de cada clase (por ejemplo, la probabilidad de que sea un pájaro y la probabilidad de que no lo sea).

2. **Impresión de la predicción**

Código Python:
```
print(f"This is a: {is_bird} ?")
```

* `is_bird` contiene la etiqueta de la predicción (por ejemplo, `'bird'` si la predicción fue que la imagen es de un pájaro). Este mensaje imprimirá algo como: **"This is a: bird ?"**, si el modelo cree que la imagen es un pájaro. **"This is a: not bird ?"**, si el modelo cree que no lo es.

3. **Impresión de la probabilidad de ser un pájaro**

Código Python:
```
print(f"Probability it's a bird: {probs[0]:.4f}")
```
* `probs[0]`: Se asume que `probs[0]` contiene la probabilidad de que la imagen sea de un pájaro (esto depende del orden en que el modelo fue entrenado).
* `{probs[0]:.4f}`: Esto formatea el valor de la probabilidad con cuatro decimales.
* El mensaje imprimirá algo como:
"Probability it's a bird: 0.9876", si el modelo está muy seguro de que es un pájaro (98.76% de probabilidad).

````python
is_bird,_,probs = learn.predict(PILImage.create('/content/bird.jpg'))
print(f"This is a: {is_bird} ?")
print(f"Probability it's a bird: {probs[0]:.4f}")
````
